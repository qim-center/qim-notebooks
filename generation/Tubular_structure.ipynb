# Tubular structure

Generates a tubular 3D synthetic volume with custom noise, gamma, and threshold properties to model tubular structures.

```python
import qim3d

vol = qim3d.generate.volume(base_shape = (200, 100, 100),
                        final_shape = (400, 100, 100),
                        noise_scale = 0.03,
                        gamma = 0.12,
                        threshold = 0.85,
                        shape = "tube",
                        )
```

```python
qim3d.viz.slices_grid(vol, n_slices=5, slice_axis=0)
```

```python
qim3d.viz.slices_grid(vol, n_slices=5, slice_axis=1)
```

```python
qim3d.viz.volumetric(vol)
```
