# Shape Generators

Generates 3D synthetic volumes of complex geometric shapes such as berries and ropes with customizable parameters and optional label segmentations.

```python
import qim3d
import numpy as np
```

### **Example 1**: Berry volume generation and drupelet label segmentation

Generates a synthetic 3D berry volume with individual drupelets and retrieves both the volume intensity array and drupelet label segmentations.

```python
# Generate a synthetic berry volume with labels
berry_vol, labels = qim3d.generate.berry(
    shape=(200, 200, 200),
    num_drupelets=80,
    seed=42,
    return_labels=True
)

# Visualize volume slices
qim3d.viz.slices_grid(berry_vol, n_slices=15)
```

```python
# Create segmentation colormap for drupelets
num_drupelets = int(np.max(labels))
cmap = qim3d.viz.colormaps.segmentation(n_labels=num_drupelets)

# Display slices of the labels
qim3d.viz.slices_grid(labels, colormap=cmap, max_value=num_drupelets)
```

### **Example 2**: Twisted rope generation and thread label segmentation

Generates a synthetic 3D twisted rope volume composed of multiple interwoven threads and retrieves both the volume intensity array and thread label segmentations.

```python
# Generate a synthetic twisted rope volume with thread labels
rope_vol, thread_labels = qim3d.generate.rope(
    shape=(300, 100, 100),
    num_threads=18,
    twist_rate=2.0,
    seed=42,
    return_labels=True
)

# Visualize volume slices
qim3d.viz.slices_grid(rope_vol, n_slices=15)
```

```python
# Create segmentation colormap for threads
num_threads = int(np.max(thread_labels))
cmap_threads = qim3d.viz.colormaps.segmentation(n_labels=num_threads)

# Display slices of the thread labels
qim3d.viz.slices_grid(thread_labels, colormap=cmap_threads, max_value=num_threads)
```
